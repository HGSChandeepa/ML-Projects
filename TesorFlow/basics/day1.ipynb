```python
# This is a constant

a = tf.constant([[[12,2,3,3] , [1,2,3,4]]] , dtype= tf.int16)
a.dtype
print(a.shape  , a.ndim)

# This is variable
b =  tf.Variable([[[12,2,3,3] , [1,2,3,4]]] , dtype= tf.int16)
a.dtype
print(a.shape  , a.ndim)

print(a[0][0])
```

```python
b[0, 0].assign([1, 2, 3, 4])
```

```python
# If we need to change some variable we need to asign
c =  tf.Variable([1])
c[0].assign(11111)
```
