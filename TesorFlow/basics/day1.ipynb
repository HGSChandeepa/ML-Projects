````python
# Create (3, 2) tensor
X = tf.constant([[1, 2],
                 [3, 4],
                 [5, 6]])

# Create another (3, 2) tensor
Y = tf.constant([[7, 8],
                 [9, 10],
                 [11, 12]])
X, Y
````

````python
X @ Y
````

Here we can see that the 3 matrix dimentions are  not OK for the multiplication , here we can do:
1. reshape()
2. we can transpose()


````python
# get the transpose of Y
Y_transpose =  tf.transpose(Y)
Y_transpose
````

````python
# now we can do the matrix multiplication
X @ Y_transpose
````

````python
# also we can use the reshape to change the shape of the tensor
Y_reshaped = tf.reshape(Y , shape=(2,3))
Y_reshaped

# now we can do the matrix multiplication
mat_mul = X @ Y_reshaped

# the dot product
dot_product = tf.tensordot(X , Y_reshaped , axes=1)


mat_mul , dot_product
````

### Create a random tensor using Randomm

````python
import numpy as np

random_tensor  = tf.constant(np.random.randint(0, 100, size=(3, 3)))
random_tensor.shape
````

### Here we can do the calculations

1. Min
2. Max
3. Mean
4. Sum
5. argmax() - find the position of the maximum element in a given tensor.
6. argmin() - find the position of the minimum element in a given tensor.

````python
min_random_tensor = tf.reduce_min(random_tensor)
max_random_tensor = tf.reduce_max(random_tensor)
mean_random_tensor = tf.reduce_mean(random_tensor)
sum_random_tensor = tf.reduce_sum(random_tensor)


# argmin and argmax
argmin_random_tensor = tf.argmin(random_tensor)
argmax_random_tensor = tf.argmax(random_tensor)

print(random_tensor)
min_random_tensor , max_random_tensor , mean_random_tensor , sum_random_tensor , argmin_random_tensor , argmax_random_tensor

````


### One-hot encoding
If you have a tensor of indicies and would like to one-hot encode it, you can use tf.one_hot().

You should also specify the depth parameter (the level which you want to one-hot encode to).

````python
# Create a list of indices
some_list = [0, 1, 2, 3]

# One hot encode them
tf.one_hot(some_list, depth=len(some_list))
````


### Using @tf.function

In your TensorFlow adventures, you might come across Python functions which have the decorator @tf.function.

### Lets understand python Decorators

Python decorators allow you to modify or extend the behavior of functions and methods without changing their actual code. When you use a Python decorator, you wrap a function with another function, which takes the original function as an argument and returns its modified version.

here is an example code:

```

def decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper

def say_whee():
    print("Whee!")

say_whee = decorator(say_whee)

---output ---
>>> from hello_decorator import say_whee

>>> say_whee()
Something is happening before the function is called.
Whee!
Something is happening after the function is called.

```




````python
# Create a simple function
def function(x, y):
  return x ** 2 + y

x = tf.constant(np.arange(0, 10))
y = tf.constant(np.arange(10, 20))
function(x, y)
````

````python
# Create the same function and decorate it with tf.function
@tf.function
def tf_function(x, y):
  return x ** 2 + y

tf_function(x, y)
````

- If you noticed no difference between the above two functions (the decorated one and the non-decorated one) you'd be right.

- Much of the difference happens behind the scenes. One of the main ones being potential code speed-ups where possible.
