## 2. Load All Three Models

We load each model and measure how long it takes to download + initialize.

```python
import time
from kittentts import KittenTTS

models = {}
model_configs = {
    "Nano int8 (15M params, ~20MB) ⭐": "KittenML/kitten-tts-nano-0.8-int8",
    "Nano fp32 (15M params, ~50MB)": "KittenML/kitten-tts-nano-0.8-fp32",
    "Micro (40M params, ~40MB)": "KittenML/kitten-tts-micro-0.8",
    "Mini (80M params, ~79MB)": "KittenML/kitten-tts-mini-0.8",
}

for label, model_id in model_configs.items():
    print(f"⏳ Loading {label}...")
    t0 = time.time()
    models[label] = KittenTTS(model_id)
    print(f"   ✅ Loaded in {time.time() - t0:.1f}s\n")

print("All models ready!")
```

## 3. Helper: Generate & Play Audio

A utility function that generates speech, measures inference time, and plays the audio inline.

```python
import IPython.display as ipd
import numpy as np

SAMPLE_RATE = 24000

def generate_and_play(model, model_label, text, voice):
    """Generate TTS audio, print timing stats, and play inline."""
    t0 = time.time()
    audio = model.generate(text, voice=voice)
    elapsed = time.time() - t0

    duration = len(audio) / SAMPLE_RATE
    rtf = elapsed / duration if duration > 0 else float('inf')

    print(f"🔊 {model_label} | Voice: {voice}")
    print(f"   Inference: {elapsed:.2f}s | Audio: {duration:.2f}s | RTF: {rtf:.2f}x")
    display(ipd.Audio(audio, rate=SAMPLE_RATE, autoplay=False))
    print()
```

## 4. Comparison: Same Text, All Models × Multiple Voices

We test a diverse set of sentences to evaluate clarity, expressiveness, and naturalness across all model sizes.

### 4a. Conversational sentence — Voice: Bella

```python
text = "Hey, I was just thinking about grabbing coffee later. Want to come with me?"

for label, model in models.items():
    generate_and_play(model, label, text, voice="Bella")
```

### 4d. News / formal — Voice: Hugo

## 5. Voice Showcase: All 8 Voices on the Mini Model

Since Mini is the largest model, let's hear what every voice sounds like at peak quality.

```python
voices = ['Bella', 'Jasper', 'Luna', 'Bruno', 'Rosie', 'Hugo', 'Kiki', 'Leo']
showcase_text = "The future of artificial intelligence is incredibly exciting, and I can't wait to see what we build together."

mini_model = models["Mini (80M params, ~79MB)"]

for voice in voices:
    generate_and_play(mini_model, "Mini", showcase_text, voice=voice)
```

## 6. Benchmark Summary

Run a quick benchmark across all models with the same text to get a clean speed comparison.

```python
import pandas as pd

bench_text = "This is a benchmark sentence to compare the inference speed of all three Kitten TTS models under identical conditions."
bench_voice = "Jasper"
results = []

for label, model in models.items():
    times = []
    for _ in range(3):  # 3 runs for stability
        t0 = time.time()
        audio = model.generate(bench_text, voice=bench_voice)
        times.append(time.time() - t0)

    avg_time = np.mean(times)
    audio_dur = len(audio) / SAMPLE_RATE

    results.append({
        "Model": label,
        "Avg Inference (s)": round(avg_time, 3),
        "Audio Duration (s)": round(audio_dur, 2),
        "RTF (lower=faster)": round(avg_time / audio_dur, 3),
    })

df = pd.DataFrame(results)
print("\n📊 Benchmark Results (3-run average, voice=Jasper)\n")
display(df)
```
